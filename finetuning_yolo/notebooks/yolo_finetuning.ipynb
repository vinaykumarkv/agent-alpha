```python
from ultralytics import YOLO
model = YOLO("yolov8n.pt")
model.train(data="data/data.yaml", epochs=20, imgsz=640, batch=8, project="finetuning_yolo/runs", name="experiment1", exist_ok=True)
metrics = model.val()
print(metrics)

results = model("data/images/test.jpg")
results[0].show()
model.export(format="pt", project="finetuning_yolo/runs", name="experiment1", exist_ok=True)

```
